```python
import pandas as pds
from scipy import stats
file = ('run17.csv')
data = pds.read_csv(file)
male = data[data['sex'] == 'M']['pace_sec']
female = data[data['sex'] == 'F']['pace_sec']
print("This dataset contains the average pace of runners who completed a 10 mile run. Enter your pace to see how you compare.")
gender = input("What gender are you? Enter 'M' for male and 'F' for female")
paceM = float(input("Enter your average pace for a 10 mile run in minutes:"))
paceS = float(input("And how many seconds:"))
pace = (paceM*60)+paceS
if gender == 'M':
    percentilePace = stats.percentileofscore(male, pace, kind='weak', nan_policy='omit').round()
if gender == 'F':
    percentilePace = stats.percentileofscore(female, pace, kind='weak', nan_policy='omit').round()
print('Your pace falls in the', 100-percentilePace, 'th Percentile')

```
